# Teste 1 — Fundamentos Matemáticos

**Contexto:** uma empresa de comércio eletrônico quer entender como os custos mensais de transporte estão variando e usar os dados dos últimos 12 meses para prever os custos dos próximos meses.

Para isso, vamos usar uma **função afim** (f(x) = ax + b), que é a função mais simples capaz de representar uma tendência de crescimento (ou queda) aproximadamente constante ao longo do tempo.


## Etapa 1 — Carregar os dados

Primeiro, vamos carregar o arquivo `base_custos_transporte.csv`, que contém duas colunas:

- `mes`: o número do mês (de 1 a 12);
- `custo_transporte`: o custo total de transporte naquele mês, em reais.

Vamos usar a biblioteca **pandas**, que serve para carregar e organizar dados de tabelas dentro do Python.


```python
# pandas será usado para carregar e organizar os dados da tabela
import pandas as pd

# Carrega o arquivo CSV com os dados de custo de transporte
dados = pd.read_csv("base_custos_transporte.csv")

# Mostra a tabela carregada, para confirmar que os dados foram lidos corretamente
dados
```

## Etapa 2 — Identificar as variáveis

Na base de dados temos duas colunas, que representam duas variáveis do problema:

**a) Variável independente:** `mes`.
É o número do mês (1, 2, 3, ..., 12). Ela é chamada de independente porque o seu valor não depende de nada além da própria contagem do tempo — o mês 5 é sempre o mês 5.

**b) Variável dependente:** `custo_transporte`.
É o custo de transporte, em reais, registrado em cada mês. Ela é chamada de dependente porque o seu valor **depende do mês** em que estamos observando: o custo muda conforme o tempo passa.

**c) O que cada variável representa no problema:**
- `mes` representa o momento no tempo (qual mês do ano de operação da empresa estamos analisando);
- `custo_transporte` representa o valor gasto pela empresa com transporte naquele mês específico, e é justamente essa grandeza que queremos entender e prever.


## Etapa 3 — Construir um gráfico

Agora vamos construir um **gráfico de dispersão** (pontos), colocando:

- o mês no eixo X;
- o custo de transporte no eixo Y.

Vamos usar a biblioteca **matplotlib**, que serve para criar gráficos a partir de dados numéricos.


```python
# matplotlib.pyplot será usado para criar os gráficos
import matplotlib.pyplot as plt

# Separamos as colunas em duas variáveis simples, para facilitar o uso
mes = dados["mes"]
custo = dados["custo_transporte"]

# Cria a figura do gráfico
plt.figure(figsize=(8, 5))

# Gráfico de dispersão: mes no eixo X, custo no eixo Y
plt.scatter(mes, custo, color="blue")

# Título e nomes dos eixos
plt.title("Custo de Transporte ao Longo dos Meses")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")

# Grade para facilitar a leitura do gráfico
plt.grid(True)

plt.show()
```

**Comportamento observado:** os pontos do gráfico mostram uma tendência praticamente contínua de **crescimento** do custo de transporte ao longo dos meses. Não há grandes saltos ou quedas bruscas — os pontos sobem de forma bastante regular, o que sugere que uma **função afim** (uma reta) pode representar bem essa tendência.


## Etapa 4 — Encontrar uma função

Queremos encontrar uma função afim:

**f(x) = ax + b**

que represente aproximadamente o comportamento dos custos, onde:

- **x** é o mês;
- **a** é o coeficiente angular (mostra o quanto o custo sobe, em média, a cada mês);
- **b** é o coeficiente linear (mostra um valor de referência inicial da reta).

### Como vamos calcular a e b

Como os pontos não estão perfeitamente alinhados em uma reta, vamos usar o método do **ajuste linear pelos mínimos quadrados**. Esse método encontra a reta que fica, em média, mais próxima de todos os pontos.

As fórmulas usadas são:

$$a = \dfrac{n \cdot \sum(xy) - \sum(x) \cdot \sum(y)}{n \cdot \sum(x^2) - \left(\sum(x)\right)^2}$$

$$b = \dfrac{\sum(y) - a \cdot \sum(x)}{n}$$

Onde:
- **n** é a quantidade de meses (pontos) que temos;
- **∑(x)** é a soma de todos os meses;
- **∑(y)** é a soma de todos os custos;
- **∑(xy)** é a soma de cada mês multiplicado pelo seu próprio custo;
- **∑(x²)** é a soma de cada mês elevado ao quadrado.

Vamos calcular cada uma dessas somas passo a passo, sem usar matrizes ou vetores — apenas somas simples.


```python
# Quantidade de meses (pontos) que temos na base
n = len(mes)

# Soma de todos os meses (soma de x)
soma_x = mes.sum()

# Soma de todos os custos (soma de y)
soma_y = custo.sum()

# Soma de cada mes multiplicado pelo seu custo (soma de x vezes y)
soma_xy = (mes * custo).sum()

# Soma de cada mes elevado ao quadrado (soma de x ao quadrado)
soma_x2 = (mes ** 2).sum()

print("n =", n)
print("soma de x =", soma_x)
print("soma de y =", soma_y)
print("soma de xy =", soma_xy)
print("soma de x ao quadrado =", soma_x2)
```

Agora vamos substituir esses valores nas fórmulas de **a** e **b**:


```python
# Cálculo do coeficiente a, substituindo os valores na fórmula
a = (n * soma_xy - soma_x * soma_y) / (n * soma_x2 - soma_x ** 2)

# Cálculo do coeficiente b, substituindo os valores na fórmula (já usando o "a" encontrado acima)
b = (soma_y - a * soma_x) / n

print("Substituindo na fórmula de a:")
print(f"a = ({n} * {soma_xy} - {soma_x} * {soma_y}) / ({n} * {soma_x2} - {soma_x}^2)")
print(f"a = {a:.2f}")

print()

print("Substituindo na fórmula de b:")
print(f"b = ({soma_y} - {a:.2f} * {soma_x}) / {n}")
print(f"b = {b:.2f}")

print()
print(f"Função encontrada: f(x) = {a:.2f}x + {b:.2f}")
```

**Função encontrada:**

$$f(x) = 1791{,}61x + 42204{,}55$$

Ou seja: **a ≈ 1791,61** e **b ≈ 42204,55**.


## Etapa 5 — Interpretar a função

**a) O que representa o coeficiente a no contexto do problema?**
O coeficiente **a (≈ 1791,61)** representa o quanto o custo de transporte aumenta, em média, a cada mês que passa. Ele é a "inclinação" da reta.

**b) O que representa o coeficiente b?**
O coeficiente **b (≈ 42204,55)** representa o valor estimado do custo de transporte no "mês zero" (ou seja, o ponto onde a reta cruzaria o eixo Y). Ele funciona como um valor de referência inicial da tendência, e não corresponde exatamente a um mês real da base.

**c) O valor de a indica crescimento ou redução dos custos?**
Como **a é positivo (≈ 1791,61)**, ele indica **crescimento**: a cada mês que passa, o custo de transporte tende a aumentar, em média, aproximadamente R$ 1.791,61 em relação ao mês anterior.


## Etapa 6 — Representar a função graficamente

Agora vamos desenhar, no mesmo gráfico:

- os **dados reais** (pontos, como na Etapa 3);
- a **reta da função encontrada** (linha de tendência calculada na Etapa 4).

Isso permite comparar visualmente o quanto a reta se aproxima dos dados reais.


```python
# Calcula o valor previsto pela função para cada mes da base (para desenhar a reta)
custo_previsto = a * mes + b

plt.figure(figsize=(8, 5))

# Pontos com os dados reais
plt.scatter(mes, custo, color="blue", label="Dados reais")

# Reta com os valores calculados pela função encontrada
plt.plot(mes, custo_previsto, color="red", label="Função encontrada (tendência)")

plt.title("Custo de Transporte: Dados Reais x Função Encontrada")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.legend()
plt.grid(True)

plt.show()
```

Como podemos ver, a reta em vermelho (função encontrada) acompanha bem de perto os pontos azuis (dados reais), confirmando que a função afim representa adequadamente a tendência de crescimento dos custos.


## Etapa 7 — Fazer previsões

Agora vamos usar a função encontrada, **f(x) = ax + b**, para estimar o custo de transporte nos meses 13, 14 e 15, que ainda não existem na base de dados.

Vamos mostrar a substituição de cada valor na fórmula, passo a passo.


```python
# Previsão para o mes 13
mes13 = 13
custo_mes13 = a * mes13 + b
print(f"f(13) = a(13) + b")
print(f"f(13) = {a:.2f}({mes13}) + {b:.2f}")
print(f"f(13) = R$ {custo_mes13:.2f}")

print()

# Previsão para o mes 14
mes14 = 14
custo_mes14 = a * mes14 + b
print(f"f(14) = a(14) + b")
print(f"f(14) = {a:.2f}({mes14}) + {b:.2f}")
print(f"f(14) = R$ {custo_mes14:.2f}")

print()

# Previsão para o mes 15
mes15 = 15
custo_mes15 = a * mes15 + b
print(f"f(15) = a(15) + b")
print(f"f(15) = {a:.2f}({mes15}) + {b:.2f}")
print(f"f(15) = R$ {custo_mes15:.2f}")
```

**Resumo das previsões:**

- Mês 13: aproximadamente **R$ 65.495,45**
- Mês 14: aproximadamente **R$ 67.287,06**
- Mês 15: aproximadamente **R$ 69.078,67**

**b) Os custos estão aumentando ou diminuindo?**

**c) Aproximadamente quanto o custo aumenta a cada mês?**
